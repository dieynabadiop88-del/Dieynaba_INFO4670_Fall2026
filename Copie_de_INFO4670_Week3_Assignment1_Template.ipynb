### Q2 · One categorical variable
**How does enrollment compare across majors?**
→ *sorted bar chart*. Column: `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
order = students["major"].value_counts().index
fig, ax = plt.subplots()
sns.countplot(data=students, y="major", order=order, color=NAVY, ax=ax)
ax.set_title("Number of students by major")
ax.set_xlabel("Number of students")
ax.set_ylabel("Major")
plt.show()
```

- **Track —**  How does enrollment compare across majors?
- **Articulate —**  Cybersecurity has by far the highest enrollment, at roughly 500 students. Business Analytics and Learning Technologies follow closely behind at around 200–250. There’s then a sharp drop-off to a cluster of much smaller categories (Cyber Security, DataScience, Information Science, Info Technology) each under 100 students.
- **Respond —** Something odd stands out: several of those small bottom categories look like duplicates of the larger ones above, just with inconsistent formatting “Cybersecurity” vs. “Cyber Security,” “Data Science” vs. “DataScience,” “Information Science” appearing twice, and “Information Technology” vs. “Info Technology.” This is likely an issue (inconsistent capitalization/spacing during data entry) rather than truly distinct majors. ### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Final GPA vs. commute miles")
ax.set_xlabel("Commute miles")
ax.set_ylabel("Final GPA")
plt.show()
```

**STAR story — Q4** *(write in your own words)*

- **Scope —**  This chart shows the relationship between commute distance and academic performance, using the commute_miles and final_gpa columns, plotted as a scatter plot with a fitted trend line.
- **Track —** Is there a relationship between how far a student commutes and their final GPA?
- **Articulate —** There’s a clear negative trend: the regression line slopes downward, from around 2.5 GPA at 0 miles down to about 1.0 GPA at 80 miles. The relationship looks weak-to-moderate rather than strong. The data is heavily clustered and scattered between 0 and 3.5 GPA for shorter commutes (0–20 miles), and gets noticeably sparser as commute distance increases past 30–40 miles, which is also where the trend line widens.
- **Respond —** Two things look odd and worth flagging. First, there’s a dense row of points sitting exactly at GPA = 0.0 across a wide range of commute distances — that looks less like real failing grades and more like a placeholder/missing-value code. Second, the sparse data at high commute distances (50–80 miles) means the downward slope out there is being driven by very few points, so the trend could be less reliable at that end. Next, I’d want to check what the 0.0 GPA values actually represent and consider excluding or investigating them before trusting the negative relationship.



### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             color=NAVY, marker="o", ax=ax)
ax.set_title("Average active minutes per student, by week")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Avg. minutes active")
plt.show()
```

**STAR story — Q5** *(write in your own words)*

- **Scope —** This chart shows average active minutes per student across each week of the semester, using the week_of_semester and avg_minutes_active columns, plotted as a time-series line chart.
- **Track —** How does student engagement change over the course of the semester?
- **Articulate —** For most of the semester, engagement is remarkably flat and stable, averaging around 210–215 active minutes per week from week 1 through week 16, with only minor week-to-week fluctuation. There’s no real upward or downward trend overall; students are consistently active throughout the term.
- **Respond —** The one thing that jumps out is a sharp, isolated V-shaped drop around week 9, where average active minutes plunges to about 95 before immediately snapping back to the normal ~210 level the following week. This is a clear outlier rather than part of any gradual trend, it doesn’t ramp down and back up, it’s a single-week spike downward. Week 9 often lines up with fall break or a midterm week on many academic calendars, so that’s my first guess, but I’d want to check the school’s calendar or look for a data-logging gap before assuming a cause.
